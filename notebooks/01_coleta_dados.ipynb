```python
# 1. Importação das Bibliotecas

import yfinance as yf
import pandas as pd
from pathlib import Path
```

```python
# 2 Definição do Ativo e Período

ticker = "^BVSP"
data_inicio = "2010-01-01"
data_fim = "2026-01-01"
```

```python
# 3. Download dos dados

df = yf.download(
    ticker,
    start=data_inicio,
    end=data_fim,
    auto_adjust=False
)

# Headers do dataset
df.head()
```

```python
# Tamanho do dataset (linhas, colunas)

df.shape
```

```python
# Informações das colunas

df.info()
```

```python
# Validação da data mais antiga e a mais recente do dataset

df.index.min(), df.index.max()
```

```python
# validação se existem valores nulos em alguma coluna

df.isna().sum()
```

```python
# Validação se existem duplicatas

df.index.duplicated().sum(),

df.duplicated().sum()
```

```python
# Visão inicial de alguns stats do dataset

df.describe()
```

```python
# Transformar retorno do yFinance em uma tabela

if isinstance(df.columns, pd.MultiIndex):
    df.columns = df.columns.get_level_values(0)

df = df.reset_index()

# Organizar a ordem das colunas
df = df[ [ "Date", "Open", "High", "Low", "Close", "Adj Close", "Volume" ] ]

# Ordenar os dados do dataset
df = df.sort_values("Date").reset_index(drop=True)

df.head()
```

```python
# Validação final dos dados

print("### Validação final ###\n")
print("Quantidade de linhas: ", len(df))
print("Data inicial: ", df["Date"].min())
print("Data final: ", df["Date"].max())
print("\nValores Ausentes:")
print(df.isna().sum())
print("\nDuplicatas: ", df.duplicated().sum())
```

```python
# Criação do diretório de saída

raw_dir = Path("data/raw")
raw_dir.mkdir(parents=True, exist_ok=True)

if raw_dir.exists() and raw_dir.is_dir():
    print(f"SUCESSO: O diretório {raw_dir} foi criado ou já existia")
else: 
    print(f"ERRO: O diretório {raw_dir} não foi encontrado")
```

```python
# Criação do arquivo e save dos dados brutos

arquivo_saida = raw_dir / "ibovespa_raw.csv"

df.to_csv(arquivo_saida, index=False)

if arquivo_saida.exists():
    print(f"Arquivo salvo em {arquivo_saida}")
else: 
    print("O arquivo não foi criado")

arquivo_saida.exists()
```

```python
print("Diretório atual:")
print(Path.cwd())

print("\nCaminho calculado:")
print(arquivo_saida)

print("\nCaminho absoluto:")
print(arquivo_saida.resolve())
```
